# gaia-sed-toolkit test case

This notebook runs the packaged SED workflow for Gaia DR3 source `5788625396770225152` with:

- `A_V = 0.0`
- `norm_band = "zmag"`

It is designed to be opened from the repository root.

```python
from pathlib import Path
import sys

repo_root = Path.cwd()
src_dir = repo_root / "src"
if str(src_dir) not in sys.path:
    sys.path.insert(0, str(src_dir))

repo_root
```

```python
from general_sed import run_single_source_sed
from IPython.display import display

SOURCE_ID = 5788625396770225152
A_V = 0.0
NORM_BAND = "zmag"

XP_DIR = Path("data/examples/xp_continuous_fits")
OUTPUT_DIR = Path("outputs/fig")
TEMP_DIR = Path("outputs/temp")
MODEL_SPEC_DIR = Path("data/models/bt-nextgen-agss2009")

if not MODEL_SPEC_DIR.exists():
    raise FileNotFoundError(
        "Expected full BT-NextGen spectra at data/models/bt-nextgen-agss2009. "
        "If you moved the grid, update MODEL_SPEC_DIR in this notebook."
    )

result = run_single_source_sed(
    source_id=SOURCE_ID,
    a_v=A_V,
    xp_dir=XP_DIR,
    output_dir=OUTPUT_DIR,
    temp_dir=TEMP_DIR,
    model_spec_dir=MODEL_SPEC_DIR,
    norm_band=NORM_BAND,
)

display(result.to_dict())
```

```python
from IPython.display import IFrame

plot_path = Path(result.plot_pdf)
print(f"Saved plot: {plot_path}")
IFrame(src=str(plot_path), width="100%", height=700)
```
